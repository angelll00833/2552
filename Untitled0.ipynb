```markdown
# Python 資料型別實作作業：聚餐費用試算器
**姓名**：林郁齡 | **學號**：D1446138

本程式包含三組測試要求的實作，依照規定：
1. 服務費 = 原始餐費 × 10%
2. 應付總額 = 原始餐費 ＋ 服務費 － 折價券
3. 每人費用 = 應付總額 ÷ 分攤人數
4. 判斷每人費用是否 > 預算 (300 元)
```

````python
# ==========================================
# 測試一：使用原始資料 (4 人、折價券 100 元)
# ==========================================

# 原始輸入資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

print("=== 測試一 ===")

# 1. 整理聚餐名稱：去除頭尾空白並轉小寫
name_clean = event_name.strip().lower()

# 2. 整理餐費與人數並轉換型別
# 移除 "NT$"、"," 及空白，並轉成 float 浮點數
meal_clean = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_num = float(meal_clean)

# 將字串人數轉成 int 整數
people_num = int(people)

# 3. 分步計算費用
service_fee = meal_num * service_rate             # 服務費
total_amount = meal_num + service_fee - discount   # 應付總額
per_person_cost = total_amount / people_num       # 每人分攤費用

# 4. 判斷是否超過預算
is_over_budget = per_person_cost > budget_per_person

# 5. 輸出費用摘要 (使用 f-string 控制小數點後兩位)
print(f"聚餐名稱：{name_clean}")
print(f"原始餐費：{meal_num:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_num} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
````

```markdown
### ❓ 問題思考與回答

#### **問題一：為什麼 "4" 要先轉成數字，才能用於本題的分攤計算？**
* **回答**：
  在 Python 中，`"4"` 的型別是**字串 (str)**，而分攤計算需要進行數學除法運算。如果直接使用字串與數字進行運算（例如：`total_amount / "4"`），會導致型別錯誤 (`TypeError`)，因為 Python 不允許將數值直接除以字串。因此，必須先使用 `int("4")` 將其轉換成**整數 (int)** 型別，才能進行正確的數學除法計算。

#### **問題二：如果先扣折價券，再依剩餘餐費計算服務費，第一組的應付總額會變成多少？為什麼和題目規定的結果不同？**
* **回答**：
  * **重新計算過程**：
    1. 扣除折價券後的餐費：$1200 - 100 = 1100$ 元。
    2. 依剩餘餐費計算服務費：$1100 \times 0.10 = 110$ 元。
    3. 新應付總額：$1100 + 110 = 1210$ 元。
  * **與原結果差異之原因**：
    原規定的應付總額是 $1220$ 元（服務費為原始餐費的 10%，即 $120$ 元）。如果先扣折價券，計算服務費的基數就會從 $1200$ 元減少為 $1100$ 元，導致服務費少收了 $10$ 元（$120 - 110 = 10$）。因為計算服務費的先後順序不同，影響了服務費的基數，所以計算出的應付總額也會有所不同。這在現實中通常取決於餐廳的財務折抵規則。
```
